```python
from ultralytics import YOLO
from google.colab import files
import os

# Load pretrained YOLO11 model
model = YOLO("yolo11n.pt")

# Train on helmet dataset
model.train(
    data="/content/helmet_dataset/data.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/helmet_training",
    name="helmet_detector",
    exist_ok=True
)

print("Training completed!")

# Locate the best trained model
best_path = "/content/helmet_training/helmet_detector/weights/best.pt"

if os.path.exists(best_path):
    print("Best model found!")
    print("Downloading best.pt to your laptop...")
    files.download(best_path)
else:
    print("ERROR: best.pt was not found!")
```

```python
from ultralytics import YOLO

# Load your trained model from the current Colab session
model = YOLO("/content/helmet_training/helmet_detector/weights/best.pt")

# Evaluate on unseen test images
metrics = model.val(
    data="/content/helmet_dataset/data.yaml",
    split="test",
    imgsz=640,
    device=0
)

print("Test evaluation completed!")
```
